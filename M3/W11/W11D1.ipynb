```python
# importiamo le variabili con le credenziali con il codice precedente, e
# poi:

import pymysql

# creiamo la variabile "connection" che gestisce la connessione a database
connection = pymysql.connect(
    host=hostname,
    user=username,
    password=password,
    database=database,
    cursorclass=pymysql.cursors.DictCursor
)
```

```python
import pandas as pd

cursor = connection.cursor()
query = "SELECT * FROM dimproduct;"
cursor.execute(query)
result = cursor.fetchall()
dimproduct = pd.DataFrame(result)

# Quanti dati ci sono in totale ? 
# Si può vedere già dall'output ovvero 606 rows × 36 columns. Oppure 
print( dimproduct.shape ) 
print(dimproduct)

# Quali sono i metadati ? 
print ( dimproduct.info() )

# Stampiamo il primo elemento e l'ultimo, e 5 elementi a caso
dimproduct.head(1)
dimproduct.tail(1)
dimproduct.sample(5)

# I colori disponibili 
colori = dimproduct["Color"].unique()

for colore in colori:
    print(colore)

# Quanto pesano in media prodotti ? Il più leggero, il più pesante

dimproduct["Weight"].mean()
dimproduct["Weight"].min()
dimproduct["Weight"].max()

# Quanti prodotti pesano più di 100
filtro100 = dimproduct["Weight"] > 100
prodotti100 = dimproduct[filtro100]
prodotti100["ProductKey"].count()

# Quanto costano in media i prodotti
dimproduct["DealerPrice"].mean()

# Quali sono i quartili dei prezzi. Dopo alcune imprecazioni ho notato che DealerPrice conteneva stringhe 'object'
dimproduct["DealerPrice"] = pd.to_numeric(dimproduct["DealerPrice"])
dimproduct["DealerPrice"].describe()

# Prezzo medio prodotti di colore blu
colors = dimproduct.groupby("Color")
colors.get_group("Blue")["DealerPrice"].mean() #oppure

filtroblu = dimproduct["Color"] == "Blue"
dimproduct[filtroblu]["DealerPrice"].mean()

# Prezzo medio prodotti di colore rosso o nero
filtroRN = ( (dimproduct["Color"] == "Blue") | (dimproduct["Color"] == "Black") )
dimproduct[filtroRN]["DealerPrice"].mean() #oppure

dimproduct[dimproduct["Color"].isin(["Blue", "Black"])]["DealerPrice"].mean()

# Prezzo massimo per i prodotti di taglia 42 e peso oltre i 10 kg
filtroTP = ( (dimproduct["Size"] == "42") & (dimproduct["Weight"] > 10) )
dimproduct[filtroTP]["DealerPrice"].max()

# nome e costo per i prodotti di taglia 42 e peso oltre i 10 kg e colore argento
filtroTPC = ( (dimproduct["Size"] == "42") & (dimproduct["Weight"] > 10) & (dimproduct["Color"] == "Silver") )
colonne = ["EnglishProductName", "StandardCost"]
dimproduct[filtroTPC].loc[:,colonne]

# visualizzare standard cost e dealer price ultimi 20 elementi del dataset
dimproduct.loc[:,colonne].tail(20) # si può osservare che lo standard cost di un prodotto tendenzialmente non varia in funzione della taglia. Fa eccezione la categoria "Bottom Bracket", in cui il costo cambia tra le varianti.
```

```python
path2 = "C:/Users/giovanni.esposito/Desktop/EPICODE/M3 PYTHON/datasets/beginner_datasets/insurance.csv"
insurance = pd.read_csv(path2)

print("Le dimensioni del data set sono (righe/colonne): ", insurance.shape)
print("---")

insurance.head(10) #per avere un'anteprima di 10 righe

insurance.columns # per i nomi colonna

# finalmente usiamo groupby
insurance.groupby("region")["charges"].mean()  #differenze significative? senza fare nessuna analisi di distribuzione, probabilità ecc. diciamo solo che a west i premi assicurativi sono mediamente più bassi che a est.
insurance.groupby("smoker")["charges"].mean() #differenze significative? i fumatori pagano giustamente molto di più, in media un premio 3,8 volte maggiore
insurance.groupby("sex")["charges"].mean() #differenze significative? uomini con premi più alti in media dell'11%

# vediamo bmi
insurance["bmi"].describe()  # i descrittori statistici di bmi 

q1 = insurance["bmi"].quantile(0.25)
q2 = insurance["bmi"].quantile(0.50)
q3 = insurance["bmi"].quantile(0.75)
q4 = insurance["bmi"].max()

insurance["Qbmi"] = pd.cut( insurance["bmi"], bins=[0, q1, q2, q3, q4], labels=["Q1", "Q2", "Q3", "Q4"] )  # la funzione pd.cut me l'ha suggerita AI

insurance.groupby("Qbmi", observed=True)["charges"].describe() # e finally vediamo describe di charges per quartili di bmi
```

```python
pathw = "C:/Users/giovanni.esposito/Desktop/EPICODE/M3 PYTHON/datasets/beginner_datasets/wine.csv"
vino = pd.read_csv(pathw)

vino.groupby("quality")["alcohol"].mean().sort_values()  # ordinando in modo crescente si vedono le differenze 
vino["alcohol"].mean()

# proviamo a esplicitare tutto in un output

media_per_qualita = vino.groupby("quality")["alcohol"].mean().sort_values()
media_totale = pd.Series({"Media_Totale": vino["alcohol"].mean()})
print(pd.concat([media_per_qualita, media_totale]))

# oppure 

media_per_qualita = vino.groupby("quality")["alcohol"].mean().sort_values()
media_totale = vino["alcohol"].mean()

confronto = pd.DataFrame(media_per_qualita)
confronto["vs_media"] = ""
for indice, riga in confronto.iterrows():
    if riga["alcohol"] > media_totale:
        confronto.loc[indice, "vs_media"] = ">media"
    else:
        confronto.loc[indice, "vs_media"] = "<media"
print(confronto)


# differenza alcol vini bianchi vs rossi
vino.groupby("type")["alcohol"].agg( ["min", "max","mean", "median", "std"] ) # differenze non troppo importanti, però chiaramente i bianchi hanno min e max più bassi, ma una media leggermente più alta così come maggiore varianza.
                                                                              # chissà che pessimi vini ci sono in questo dataset

# differenza per pH
vino.groupby("pH")["alcohol"].agg( ["min", "max","mean", "median", "std"] )  # cambia tutto perchè raggruppiamo per tanti valori diversi di pH. Comunque è abbastanza evidente che all'aumentare del pH aumenta la gradazione alcolica (c'è correlazione positiva)

# differenza per solfati
vino.groupby("sulphates")["alcohol"].agg( ["min", "max","mean", "median", "std"] ) # per i solfati c'è correlazione negativa
```

```python
employees_df = pd.DataFrame({
    'employee_id': [101, 102, 103, 104, 105],
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Emma'],
    'department_id': [1, 2, 1, 2, 3]
})

departments_df = pd.DataFrame({
    'department_id': [1, 2, 3],
    'department_name': ['HR', 'IT', 'Finance'],
    'location': ['New York', 'San Francisco', 'Chicago']
})

# mergiamo con merge
merge = departments_df.merge(employees_df, on="department_id")
merge

# joiniamo con join
employees_df = employees_df.set_index("department_id")
departments_df = departments_df.set_index("department_id")
join = departments_df.join(employees_df)
join

# la differenza è che nella tabella con join il department_id è l'indice mentre nella tabella con merge abbiamo l'indice di default 0, 1, 2... 
# In aggiunta merge di default fa inner join, join di default fa left join
```
