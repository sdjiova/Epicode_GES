```python
path2 = "C:/Users/giovanni.esposito/Desktop/EPICODE/M3 PYTHON/datasets/beginner_datasets/insurance.csv"
insurance = pd.read_csv(path2)

print("Le dimensioni del data set sono (righe/colonne): ", insurance.shape)
print("---")

insurance.head(10) #per avere un'anteprima di 10 righe

sw = "southwest"
se = "southeast"
nw = "northwest"
ne = "northeast"

swe = insurance["region"] == sw
sea = insurance["region"] == se
nwe = insurance["region"] == nw
nea = insurance["region"] == ne

media_swe = insurance[swe]["charges"].mean()
media_sea = insurance[sea]["charges"].mean()
media_nwe = insurance[nwe]["charges"].mean()
media_nea = insurance[nea]["charges"].mean()

print(
    f"La media di charges nella regione {sw} è pari a {media_swe}\n"
    f"La media di charges nella regione {se} è pari a {media_sea}\n"
    f"La media di charges nella regione {nw} è pari a {media_nwe}\n"
    f"La media di charges nella regione {ne} è pari a {media_nea}"
)
print("---")

smoker1 = insurance["smoker"] == "yes"
smoker0 = insurance["smoker"] == "no"

media_smoker1 = insurance[smoker1]["charges"].mean()
media_smoker0 = insurance[smoker0]["charges"].mean()

print(
    f"La media di charges per i fumatori è pari a {media_smoker1}\n"
    f"La media di charges per i non fumatori è pari a {media_smoker0}"
)

print("---")

sexm = insurance["sex"] == "male"
sexf = insurance["sex"] == "female"

media_sexm = insurance[sexm]["charges"].mean()
media_sexf = insurance[sexf]["charges"].mean()

print(
    f"La media di charges per il sesso maschile è pari a {media_sexm}\n"
    f"La media di charges per il sesso femminile è pari a {media_sexf}"
)

insurance.describe() #Dal describe è possibile veder i descrittori statistici di BMI

print("---")

q1 = insurance["bmi"].quantile(0.25)
q2 = insurance["bmi"].quantile(0.50)
q3 = insurance["bmi"].quantile(0.75)

fascia1 = insurance[insurance["bmi"] <= q1]
fascia2 = insurance[(insurance["bmi"] > q1) & (insurance["bmi"] <= q2)]
fascia3 = insurance[(insurance["bmi"] > q2) & (insurance["bmi"] <= q3)]
fascia4 = insurance[insurance["bmi"] > q3]

min1 = fascia1["charges"].min()
media1 = fascia1["charges"].mean()
max1 = fascia1["charges"].max()

min2 = fascia2["charges"].min()
media2 = fascia2["charges"].mean()
max2 = fascia2["charges"].max()

min3 = fascia3["charges"].min()
media3 = fascia3["charges"].mean()
max3 = fascia3["charges"].max()

min4 = fascia4["charges"].min()
media4 = fascia4["charges"].mean()
max4 = fascia4["charges"].max()

print("Fascia BMI <=", q1, "| min:", min1, "| media:", media1, "| max:", max1)
print("Fascia BMI", q1, "-", q2, "| min:", min2, "| media:", media2, "| max:", max2)
print("Fascia BMI", q2, "-", q3, "| min:", min3, "| media:", media3, "| max:", max3)
print("Fascia BMI >", q3, "| min:", min4, "| media:", media4, "| max:", max4)

#Commento: una follia fare questa roba senza raggruppamenti!
```

```python
path3 = "C:/Users/giovanni.esposito/Desktop/EPICODE/M3 PYTHON/datasets/beginner_datasets/pokemon.csv"
pokemon = pd.read_csv(path3)

print("Le dimensioni del data set sono (righe/colonne): ", pokemon.shape)
print("---")

print(pokemon.columns)
print("---")

pokemon.head(20) #per avere un'anteprima di 10 righe

# La colonna '#' non coincide con l'indice automatico di Pandas.
# Si può usare come indice anche se non è consigliabile. 
# Difatti alcuni valori si ripetono e come si può evincere di seguito, indice di Pandas e i valori di '#' non coincidono:
pokemon["#"].value_counts()

pokemoni = pokemon.set_index("#")
pokemoni

filtro_legendary = pokemoni["Legendary"] == True
legendary = pokemoni[filtro_legendary]
pokemon_legendary = legendary["Name"].unique()

print(pokemon_legendary) #Questa è la lista dei pokemon leggendaree
print("---")

filtro_grass = legendary["Type 1"] == "Grass"
legendary_grass = legendary[filtro_grass]
pokemon_legendary_grass = legendary_grass["Name"].unique()

print(pokemon_legendary_grass) #Questa è la lista dei pokemon leggendaree grass 
print("---")

filtro_ice_o_fire = (legendary["Type 1"] == "Ice") | (legendary["Type 1"] == "Fire")
legendary_ice_o_fire = legendary[filtro_ice_o_fire]
pokemon_legendary_ice_o_fire = legendary_ice_o_fire["Name"].unique()

print(pokemon_legendary_ice_o_fire)
print("---")

pokemonin = pokemon.set_index("Name")
pokemonin

filtro_final = (pokemonin["Generation"] == 1) & (pokemonin["Attack"] > 50) & (pokemonin["HP"] < 60)
pokemonin_final = pokemonin[filtro_final]
vediamoli_final = pokemonin_final.index.unique()

print(vediamoli_final) #Questa è la lista dei pokemon della prima generazione con attacco maggiore di 50 e HP minore di 60
```
