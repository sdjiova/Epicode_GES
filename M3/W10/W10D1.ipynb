```python
vendite = [2000, 5500, 7200, 4320, 1280, 1900,
           2500, 3900, 6410, 8150, 7100, 5350]

vendite = pd.Series(vendite)

print("Il valore massimo è ", vendite.max())
print("Il valore minimo è ", vendite.min())
print("---")

filtro1 = vendite > 5000
print("Di seguito i valori maggiori di 5000", "\n", vendite[filtro1])
print("---")
print("I valori maggiori di 5000 sono ", len(vendite[filtro1]))
print("---")

filtro2 = vendite < 3000
print("Di seguito i valori minori di 3000", "\n", vendite[filtro2])
print("---")

print("Il valore medio delle vendite è pari a ", vendite.mean())
```

```python
input_numeri = [20, 345, 2312, 2312, 345, 2, 5, 100, 98]
input_numeri = pd.Series(input_numeri)
ordinato = input_numeri.sort_values()

if ordinato.size > 3:
    print(ordinato[-3:])
else:
    print("La lista ha meno di quattro numeri quindi questi sono già i numeri più grandi ", ordinato)

# In questo caso se sono presenti numeri uguali nella lista, l'output restituirà i numeri più grandi anche se duplicati.

# Se volessi invece i primi tre valori unici:
print("---")

input_numeri = list({20, 345, 2312, 2312, 345, 2, 5, 100, 98})
input_numeri = pd.Series(input_numeri)
ordinato = input_numeri.sort_values()

if ordinato.size > 3:
    print(ordinato[-3:])
else:
    print("La lista ha meno di quattro numeri quindi questi sono già i numeri più grandi ", ordinato)
```

```python
guadagni = [100, 90, 70, 40, 50, 80, 90, 120, 80, 20, 50, 50]

mesi = ["Gennaio", "Febbraio", "Marzo", "Aprile", "Maggio", "Giugno",
        "Luglio", "Agosto", "Settembre", "Ottobre", "Novembre", "Dicembre"]

guadagni = pd.Series(guadagni)

for indice, guadagno in guadagni.items(): 
    if indice == 0:
        print("Nel mese di ", mesi[indice], "il guadagno è stato pari a ", guadagno  )
    else: 
        guadagni_prec = guadagni[:indice]
        media = guadagni_prec.mean()
        risultato = ""
        if guadagno > media: 
          risultato = "maggiore"  
        else:
          risultato = "minore"
        print("Nel mese di ", mesi[indice], "il guadagno è stato pari a ", guadagno, "La media dei mesi precedenti è pari a ", media, ". Il guadagno attuale è ", risultato)
```
