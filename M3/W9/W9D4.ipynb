```python
input_utente = input("Inserisci il testo ")
n = len(input_utente)
output = ""
if n >= 8:
    output = input_utente[0:3] + "..." + input_utente[-3:]
elif 6 <= n <= 7:
    output = input_utente[0:2] + "..." + input_utente[-2:]
elif 4 <= n <= 5:
    output = input_utente[0:1] + "..." + input_utente[-1:]
else:
    print("Il testo inserito è troppo breve!")
print(output)
```

```python
input_utente = int(input("Inserisci un numero intero "))
output = []
divisore = 2

while input_utente > 1:
    while input_utente % divisore == 0:
        output.append(divisore)
        input_utente = input_utente // divisore
    divisore += 1

print(output)
    
```
