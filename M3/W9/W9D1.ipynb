# Algoritmo per un risotto ai funghi perfetto

**1. Verifica la presenza di tutti gli ingredienti:**
riso carnaroli, funghi porcini freschi (lavati), funghi secchi, burro di qualità, olio evo, cipolla, aglio, parmigiano 24-30 mesi, vino bianco, prezzemolo, pepe.

**2. Verifica la presenza degli strumenti:**
padella piatta per scottare, pentola per il brodo, tegame largo e alto per il risotto, tagliere, coltelli da cucina, mestoli e leccapentole.

**3. Prepara gli ingredienti e apponili in appositi recipienti:**
- 3.1 Affetta i funghi freschi seguendo la funzione - 100 g * N Persone
- 3.2 Gratta il parmigiano seguendo la funzione - 20 g * N Persone
- 3.3 Pesa il riso carnaroli seguendo la funzione - 80g * N Persone
- 3.4 Pesa il burro seguendo la funzione - 10g * N Persone
- 3.5 Affetta la cipolla seguendo la funzione - 1/4 cipolla * N Persone
- 3.6 Sbuccia un aglio
- 3.7 Affetta un ciuffo di prezzemolo

**4. Esegui la cottura dei funghi freschi:**
- 4.1 Nella padella per scottare inserisci olio evo, aglio, prezzemolo
- 4.2 Accendi il fuoco, fiamma media
- 4.3 Appena gli ingredienti del punto 3.1 soffriggono inserire funghi freschi
- 4.4 Mescola gli ingredienti due volte ogni 30 secondi
- 4.5 Esegui il test - i funghi porcini sono dorati? Se TRUE → spegni; Se FLASE → ritorna al punto 4.4

**5. Esegui la preparazione del brodo:**
- 5.1 In una pentola inserisci acqua secondo la funzione - 150 ml * N Persone
- 5.2 Accendi il fuoco, fiamma vivace
- 5.3 Appena l'acqua bolle, inserisci i funghi secchi e abbassa a fiamma media
- 5.4 Lascia sobbollire per 20 minuti
- 5.5 Filtra il brodo con un colino e tieni in caldo a fuoco minimo
- 5.6 Recupera i funghi secchi reidratati e tienili da parte

**6. Esegui la tostatura del riso:**
- 6.1 Nel tegame inserisci una noce di burro e la cipolla affettata
- 6.2 Accendi il fuoco, fiamma media
- 6.3 Soffriggi la cipolla finché diventa trasparente (~3 minuti)
- 6.4 Inserisci il riso e mescola continuamente per 2 minuti
- 6.5 Esegui il test - il riso è traslucido ai bordi? Se TRUE → vai al 6.6; Se FALSE → ritorna al 6.4
- 6.6 Sfuma con il vino bianco e mescola fino a completo assorbimento

**7. Esegui la cottura del risotto:**
- 7.1 Aggiungi un mestolo di brodo caldo e mescola
- 7.2 Esegui il test - il brodo è stato assorbito? Se TRUE → vai al 7.3; Se FALSE → attendi 1 minuto e ritorna al 7.2
- 7.3 Ripeti il punto 7.1 finché il riso non è al dente (~12 minuti totali)
- 7.4 Aggiungi i funghi secchi reidratati recuperati al punto 5.6
- 7.5 Mescola e lascia insaporire per 1 minuto
- 7.6 Spegni il fuoco

**8. Esegui la mantecatura (a fuoco spento):**
- 8.1 Aggiungi il burro rimasto freddo e il parmigiano grattugiato
- 8.2 Esegui il gesto dell'onda: impugna il tegame con entrambe le mani e spingi-tira ritmicamente avanti e indietro, facendo scivolare il riso e incorporando aria fino a ottenere una consistenza cremosa e fluida
- 8.3 Esegui il test - il risotto scorre fluidamente "a onda" se inclini il tegame? Se `True` → vai al 9; Se `False` → ritorna al 8.2

**9. Impiatta e servi:**
- 9.1 Versa il risotto nei piatti
- 9.2 Aggiungi i funghi freschi scottati al punto 4 come guarnizione
- 9.3 Aggiungi una spolverata di prezzemolo e pepe a piacere
- 9.4 Servi immediatamente

# Ricaricare qualsiasi cellulare 

**1. Verifica la presenza degli sturmenti di lavoro:**
Cellulare, alimentatore con attacco usb-c, alimentatore con attacco usb vecchio, cavo usb-c - usb-c, cavo usb vecchio - usb-c, cavo usb vecchio - Lightning, cavo usb-c - Lightning

**2. Verifica il modello di cellulare**
Android, Iphone con attacco Lightning, Iphone con attacco usb-c

**3. Selezionare il giusto set di carica**
- 3.1 IF il cellulare è un un Android THEN alimentatore con attacco usb-c + cavo usb-c - usb-c OR alimentatore con attacco usb vecchio + cavo usb vecchio - usb-c
- 3.2 IF il cellulare è un Iphone con attacco Lightning THEN alimentatore con attacco usb-c + cavo usb-c - Lightning OR alimentatore con attacco usb vecchio + cavo usb vecchio - Lightning
- 3.3 IF il cellulare è un Iphone con attacco usb-c THEN alimentatore con attacco usb-c + cavo usb-c - usb-c OR alimentatore con attacco usb vecchio + cavo usb vecchio - usb-c

**4. Collega il set di carica al cellulare:**
- 4.1 Inserisci il cavo nell'alimentatore
- 4.2 Inserisci l'alimentatore nella presa elettrica
- 4.3 Collega il cavo al cellulare
- 4.4 Esegui il test - il cellulare mostra il simbolo di ricarica? IF TRUE → ricarica in corso; IF FALSE → ritorna al punto 3 e verifica il set selezionato

**5. Attendi il completamento della ricarica:**
- 5.1 Esegui il test - la batteria è al 100%? IF TRUE → vai al 6; IF FALSE → attendi e ritorna al 5.1

**6. Disconnetti il set di carica:**
- 6.1 Scollega il cavo dal cellulare
- 6.2 Rimuovi l'alimentatore dalla presa elettrica
- 6.3 Riponi il set di carica


# Innaffiare correttamente le piante

**1. Verifica la presenza degli strumenti di lavoro:**
Piante da innaffiare, annaffiatoio, sottovaso, acqua, dito per testare il terreno

**2. Verifica il tipo di pianta**
Pianta grassa/succulenta, pianta tropicale, pianta da esterno stagionale

**3. Verifica il fabbisogno idrico della pianta**
- 3.1 IF la pianta è una pianta grassa/succulenta THEN innaffia ogni 15-20 giorni con poca acqua
- 3.2 IF la pianta è una pianta tropicale THEN innaffia ogni 3-4 giorni con acqua moderata
- 3.3 IF la pianta è una pianta da esterno stagionale THEN innaffia ogni 1-2 giorni in estate, ogni 4-5 giorni in primavera/autunno

**4. Verifica lo stato del terreno**
- 4.1 Inserisci il dito nel terreno fino al secondo nodo (~2 cm)
- 4.2 Esegui il test - il terreno è asciutto? Se TRUE → vai al 5; Se FALSE → non innaffiare, ritorna al 4.1 il giorno successivo

**5. Prepara l'acqua**
- 5.1 Riempi l'annaffiatoio con acqua a temperatura ambiente
- 5.2 Esegui il test - l'acqua è fredda di frigorifero o bollente? Se TRUE → attendi che raggiunga temperatura ambiente e ritorna al 5.2; Se FALSE → vai al 6

**6. Esegui l'innaffiatura**
- 6.1 Posiziona il sottovaso sotto la pianta
- 6.2 Versa l'acqua lentamente alla base della pianta, evitando foglie e fusto
- 6.3 Continua a versare finché l'acqua non fuoriesce nel sottovaso
- 6.4 Attendi 10 minuti
- 6.5 Esegui il test - nel sottovaso è rimasta acqua in eccesso? Se TRUE → rimuovi l'acqua in eccesso dal sottovaso; Se FALSE → vai al 7

**7. Verifica il risultato**
- 7.1 Esegui il test - il terreno è uniformemente umido in superficie? Se TRUE → innaffiatura completata; Se `False` → ritorna al 6.2 e aggiungi acqua nelle zone asciutte

**8. Riponi gli strumenti**
- 8.1 Svuota e pulisci l'annaffiatoio
- 8.2 Riponi l'annaffiatoio in luogo asciutto
